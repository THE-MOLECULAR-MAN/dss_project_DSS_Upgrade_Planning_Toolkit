# Great for writing the get_read_schema method

```python

DATASET_NAME = 'llmmodelsbyconnection_live'

import dataiku
from dataikuapi.dss.dataset import DSSDatasetSettings
from pprint import pprint

# Scope: Current Project
client = dataiku.api_client()
project = client.get_default_project()
dataset = project.get_dataset(DATASET_NAME)

schema = dataset.get_schema()

# check if it was userModified
has_at_least_one_meaning = any("meaning" in d for d in schema)

if not has_at_least_one_meaning:
    print("\n\n=================== WARNING, no meanings defined! ==================\n")

schema.pop("userModified", None)

print("""
    def get_read_schema(self):
        \"\"\"Returns the read schema for TBD\"\"\"
        # Data types: https://developer.dataiku.com/latest/api-reference/python/datasets.html#dataiku.core.dataset.Schema
        # Meanings: Text, JSONArrayMeaning, Email, Boolean, DatetimeNoTz, Date, FreeText, LongMeaning""")
print("        return ", end="")
pprint(schema)
print("\n\n\n")
```
