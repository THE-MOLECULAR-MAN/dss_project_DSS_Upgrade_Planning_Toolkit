```python
import dataiku
from dataikuapi.dss.recipe import DSSRecipe

def get_python_recipe_code_env(recipe: DSSRecipe) -> str:
    """
    Retrieves the name of the code environment used by a Python recipe.
    
    Args:
        recipe (dataikuapi.dss.recipe.DSSRecipe): The handle to the recipe object.
        
    Returns:
        str: The name of the code environment if the recipe is a 'python' recipe.
             Returns "" (empty string) if it is not a Python recipe or if no
             specific environment is configured (implying default).
    """
    # Retrieve the settings for the provided recipe handle
    settings = recipe.get_settings()
    
    # Check if the recipe type is explicitly a Python code recipe
    if settings.type == 'python':
        # Retrieve the code env name.
        # get_code_env_name() returns None if using the project default or inheritance
        env_name = settings.get_code_env_name()
        
        if env_name:
            return env_name
            
    return ""

# --- Usage Example ---
if __name__ == "__main__":
    # Standard boilerplate based on your assumptions
    client = dataiku.api_client()
    project = client.get_default_project()
    
    # Example: Replace 'compute_data_python' with your actual recipe name
    try:
        my_recipe = project.get_recipe("compute_data_python")
        env_name = get_python_recipe_code_env(my_recipe)
        print(f"Code Environment: '{env_name}'")
    except Exception as e:
        print(f"Could not retrieve recipe or environment: {e}")
```
