```python
import dataiku

from pprint import pprint as pp
#from json import dumps as jd

PLUGIN_IDs = ['model-stress-test', 'default-samples']

client = dataiku.api_client()

for p_id in PLUGIN_IDs:
    plugin_handle = client.get_plugin(p_id)
    settings_raw = plugin_handle.get_settings().get_raw()
    print(f"\n------------ {p_id} -------------")
    pp(settings_raw)
    print("\n")
```
